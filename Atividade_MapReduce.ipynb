#I. O que é o MapReduce

O MapReduce é um paradigma de programação e um modelo de processamento distribuído projetado para processar e gerar grandes volumes de dados em paralelo usando clusters de hardware comercial. Ele divide a computação em duas fases principais: o Map, que filtra e transforma os dados em pares chave-valor, e o Reduce, que agrega e resume esses resultados. No ecossistema Hadoop, o MapReduce atua como reponsável pelo processamento em lote que gerencia a distribuição das tarefas e a tolerância a falhas de forma transparente para o desenvolvedor.

Como descreve White (2015, p. 30) em seu guia definitivo sobre a tecnologia:

"MapReduce is a programming model for processing large data sets with a parallel, distributed algorithm on a cluster." (O MapReduce é um modelo de programação para processar grandes conjuntos de dados com um algoritmo paralelo e distribuído em um cluster).

Complementando a visão clássica da computação distribuída, o conceito fundamental por trás dessa arquitetura é a proximidade entre dados e processamento (data locality). Conforme apontam os criadores do Hadoop, White e outros especialistas da área (veja, por exemplo, o clássico artigo de Dean e Ghemawat, 2008, p. 108):

"Computation is moved to the data instead of moving the data to the computation." (A computação é movida para os dados em vez de mover os dados para a computação).

Esse paradigma tem como objetivo minimizar o tráfego de rede e garantir alta escalabilidade para analises de Big Data.

##Funções utilizadas

```python
def map_coluna(nome_arquivo_entrada, nome_arquivo_saida, indice_coluna: int):
  """Lê um arquivo delimitado por vírgulas, extrai uma coluna e gera um formato de mapeamento.

    A função consome um arquivo CSV/TXT de entrada, descarta o cabeçalho, extrai o valor
    presente no índice da coluna informado e grava no arquivo de saída o par (valor, 1) .

    Args:
        nome_arquivo_entrada: Caminho para o arquivo de dados.
        nome_arquivo_saida : Caminho para o arquivo resultante.
        indice_coluna (int): Índice numérico da coluna desejada.
    """

  arquivo_entrada = open(nome_arquivo_entrada, 'r')
  arquivo_saida = open(nome_arquivo_saida, 'w')
  primeira_linha = arquivo_entrada.readline()

  for linha in arquivo_entrada:
    arquivo_saida.write(('%s\t%s\n'% (linha.split(',')[indice_coluna], 1)))

  arquivo_entrada.close()
  arquivo_saida.close()
```

```python
def shuffle(nome_arquivo_entrada, nome_arquivo_saida):
  """Realiza a fase de shuffle agrupando os valores associados a cada chave.

    Lê um arquivo contendo pares de chave e valor,
    agrupa os valores em listas utilizando um dicionário padrão e grava
    o resultado estruturado no arquivo de saída.

    Args:
        nome_arquivo_entrada: Caminho do arquivo com os pares chave-valor.
        nome_arquivo_saida: Caminho do arquivo onde o dicionário agrupado será salvo.

    """
  resultado_intermediario = defaultdict(list)

  with open(nome_arquivo_entrada) as arquivo_entrada:
    for linha in arquivo_entrada:
      chave, valor = linha.split('\t')
      resultado_intermediario[chave].append(int(valor))

  resultado = dict(resultado_intermediario)
  arquivo_shuffle = open(nome_arquivo_saida, 'w')
  arquivo_shuffle.write(str(resultado))
  arquivo_shuffle.close()
```

```python
def reducer(nome_arquivo_entrada, nome_arquivo_saida, funcao_reduce):
  """Realiza a fase de reduce somando os valores agrupados por chave.

    Lê o arquivo gerado pelo shuffle contendo o dicionário agrupado,
    soma as listas de ocorrências para cada chave e grava o resultado final.

    Args:
        nome_arquivo_entrada: Caminho do arquivo com o dicionário do shuffle.
        nome_arquivo_saida: Caminho do arquivo com a contagem final consolidada.
    """
  arquivo = open(nome_arquivo_entrada,'r')
  conteudo_entrada = arquivo.read()
  dict_entrada = eval(conteudo_entrada)
  arquivo.close()
  dict_result = {}

  for chave, lista in dict_entrada.items():
    dict_result[chave] = reduce(funcao_reduce, lista)

  arquivo_reduce = open(nome_arquivo_saida, 'w')

  for item  in sorted(dict_result, key = dict_result.get, reverse = True):
    arquivo_reduce.write(item + ' ' + str(dict_result[item]))
    arquivo_reduce.write("\n")

  arquivo_reduce.close()
```

```python
def somar_elementos(x,y):
  """Soma dois elementos numéricos."""
  return x+y
```

Acima temos as duas funções que represetam as duas etapas do MapReduce. A primeira mapeia a coluna fazendo com que cada linha se torne um par de chave, valor. Ou seja cada linha representa uma chave e na frente o valor 1. A segunda é a etapa shuffle, que agrupa o que foi mapeado. A terceira realiza a etapa final reduzindo os valores gerados pela etapa shuffle e aplicando alguma operação de agregação (soma, contagem, ou filtro).

##3. Tarifa média cobrada nas viagens.

A estratégia para responder essa pergunta é parecida com a segunda questão. Porém nesta, não passaremos uma coluna para ser a chave no par chave, valor, mas sim uma palavra que será fixa para todos os mapeamentos. A função shuffle continuará a mesma da questão anterior. Já a função reducer será alterada para calcular a média da tarifa.

```python
def map_tarifa_media(nome_arquivo_entrada, nome_arquivo_saida, indice_valor:int):
  """Lê um arquivo delimitado por vírgulas, extrai uma coluna e gera um formato de mapeamento.

    A função consome um arquivo CSV/TXT de entrada, descarta o cabeçalho, extrai o valor
    presente no índice da coluna informado e grava no arquivo de saída o par (valor, 1) .

    Args:
        nome_arquivo_entrada: Caminho para o arquivo de dados.
        nome_arquivo_saida : Caminho para o arquivo resultante.
        indice_valor (int): Índice númerico da coluna que será o valor.
    """

  arquivo_entrada = open(nome_arquivo_entrada, 'r')
  arquivo_saida = open(nome_arquivo_saida, 'w')
  primeira_linha = arquivo_entrada.readline()

  for linha in arquivo_entrada:
    arquivo_saida.write(('%s\t%s\n'% ('tarifa',
                                      linha.split(',')[indice_valor])))

  arquivo_entrada.close()
  arquivo_saida.close()
```

```python
def shuffle_tarifa_media(nome_arquivo_entrada, nome_arquivo_saida):
  """Realiza a fase de shuffle agrupando os valores associados a cada chave.

    Lê um arquivo contendo pares de chave e valor,
    agrupa os valores em listas utilizando um dicionário padrão e grava
    o resultado estruturado no arquivo de saída.

    Args:
        nome_arquivo_entrada: Caminho do arquivo com os pares chave-valor.
        nome_arquivo_saida: Caminho do arquivo onde o dicionário agrupado será salvo.

    """
  resultado_intermediario = defaultdict(list)

  with open(nome_arquivo_entrada) as arquivo_entrada:
    for linha in arquivo_entrada:
      chave, valor = linha.split('\t')
      resultado_intermediario[chave].append(float(valor))

  resultado = dict(resultado_intermediario)
  arquivo_shuffle = open(nome_arquivo_saida, 'w')
  arquivo_shuffle.write(str(resultado))
  arquivo_shuffle.close()
```

```python
def reducer_tarifa_media(nome_arquivo_entrada, nome_arquivo_saida, funcao_reduce):
  """Realiza a fase de reduce somando os valores agrupados por chave.

    Lê o arquivo gerado pelo shuffle contendo o dicionário agrupado,
    soma as listas de ocorrências para cada chave e grava o resultado final.

    Args:
        nome_arquivo_entrada: Caminho do arquivo com o dicionário do shuffle.
        nome_arquivo_saida: Caminho do arquivo com a contagem final consolidada.
    """
  arquivo = open(nome_arquivo_entrada,'r')
  conteudo_entrada = arquivo.read()
  dict_entrada = eval(conteudo_entrada)
  arquivo.close()
  dict_result = {}

  for chave, lista in dict_entrada.items():
    soma = reduce(funcao_reduce, lista)
    dict_result[chave] = soma / float(len(lista) if len(lista) > 0 else 0)

  arquivo_reduce = open(nome_arquivo_saida, 'w')

  for item  in sorted(dict_result, key = dict_result.get, reverse = True):
    arquivo_reduce.write(f"{item} {dict_result[item]:.2f}\n")
    arquivo_reduce.write("\n")

  arquivo_reduce.close()
```

##4. Data e hora em que foi feita a viagem mais longa.

```python
def map_coluna(nome_arquivo_entrada, nome_arquivo_saida, indice_data: int, indice_distancia: int):
    """Extrai a data/hora e a distância da viagem."""
    arquivo_entrada = open(nome_arquivo_entrada, 'r')
    arquivo_saida = open(nome_arquivo_saida, 'w')
    primeira_linha = arquivo_entrada.readline()

    for linha in arquivo_entrada:
        if not linha.strip():
            continue
        colunas = linha.strip().split(',')
        data_hora = colunas[indice_data]
        distancia = colunas[indice_distancia]
        arquivo_saida.write(f"maior\t{distancia},{data_hora}\n")

    arquivo_entrada.close()
    arquivo_saida.close()
```

```python
def shuffle(nome_arquivo_entrada, nome_arquivo_saida):
    resultado_intermediario = defaultdict(list)

    with open(nome_arquivo_entrada) as arquivo_entrada:
        for linha in arquivo_entrada:
            if not linha.strip():
                continue
            chave, valor = linha.strip().split('\t')
            resultado_intermediario[chave].append(valor)

    resultado = dict(resultado_intermediario)
    arquivo_shuffle = open(nome_arquivo_saida, 'w')
    arquivo_shuffle.write(str(resultado))
    arquivo_shuffle.close()
```

```python
def reducer(nome_arquivo_entrada, nome_arquivo_saida):
    arquivo = open(nome_arquivo_entrada, 'r')
    conteudo_entrada = arquivo.read()
    dict_entrada = eval(conteudo_entrada)
    arquivo.close()

    maior_distancia = -1.0
    data_maior_viagem = ""

    for chave, lista in dict_entrada.items():
        for item in lista:
            # Separa a distância da data/hora que vieram juntas no valor
            distancia_str, data_hora = item.split(',')
            distancia = float(distancia_str)

            if distancia > maior_distancia:
                maior_distancia = distancia
                data_maior_viagem = data_hora

    arquivo_reduce = open(nome_arquivo_saida, 'w')
    arquivo_reduce.write(f"Distancia {maior_distancia} | Data e Hora: {data_maior_viagem}\n")
    arquivo_reduce.close()
```

##5. Quantidade de viagens por hora.

```python
def map_coluna(nome_arquivo_entrada, nome_arquivo_saida, indice_coluna: int):
  """Lê a data/hora, extrai apenas a hora e gera o par (hora, 1)."""
  arquivo_entrada = open(nome_arquivo_entrada, 'r')
  arquivo_saida = open(nome_arquivo_saida, 'w')
  primeira_linha = arquivo_entrada.readline() # descarta cabeçalho

  for linha in arquivo_entrada:
    colunas = linha.strip().split(',')
    data_hora = colunas[indice_coluna]

    try:
      # Pega o texto, separa data e hora pelo espaço e pega apenas a hora (antes dos dois pontos)
      hora = data_hora.strip().split(' ')[1].split(':')[0]
      arquivo_saida.write(f"{hora}\t1\n")
    except (IndexError, AttributeError):
      # Ignora linhas que venham com formato de data/hora inválido ou vazio
      continue

  arquivo_entrada.close()
  arquivo_saida.close()
```

```python
def shuffle(nome_arquivo_entrada, nome_arquivo_saida):
  """Realiza a fase de shuffle agrupando os valores associados a cada chave.

    Lê um arquivo contendo pares de chave e valor,
    agrupa os valores em listas utilizando um dicionário padrão e grava
    o resultado estruturado no arquivo de saída.

    Args:
        nome_arquivo_entrada: Caminho do arquivo com os pares chave-valor.
        nome_arquivo_saida: Caminho do arquivo onde o dicionário agrupado será salvo.

    """
  resultado_intermediario = defaultdict(list)

  with open(nome_arquivo_entrada) as arquivo_entrada:
    for linha in arquivo_entrada:
      chave, valor = linha.split('\t')
      resultado_intermediario[chave].append(int(valor))

  resultado = dict(resultado_intermediario)
  arquivo_shuffle = open(nome_arquivo_saida, 'w')
  arquivo_shuffle.write(str(resultado))
  arquivo_shuffle.close()
```

```python
def reducer(nome_arquivo_entrada, nome_arquivo_saida, funcao_reduce):
  """Realiza a fase de reduce somando os valores agrupados por chave.

    Lê o arquivo gerado pelo shuffle contendo o dicionário agrupado,
    soma as listas de ocorrências para cada chave e grava o resultado final.

    Args:
        nome_arquivo_entrada: Caminho do arquivo com o dicionário do shuffle.
        nome_arquivo_saida: Caminho do arquivo com a contagem final consolidada.
    """
  arquivo = open(nome_arquivo_entrada,'r')
  conteudo_entrada = arquivo.read()
  dict_entrada = eval(conteudo_entrada)
  arquivo.close()
  dict_result = {}

  for chave, lista in dict_entrada.items():
    dict_result[chave] = reduce(funcao_reduce, lista)

  arquivo_reduce = open(nome_arquivo_saida, 'w')

  for item  in sorted(dict_result, key = dict_result.get, reverse = True):
    arquivo_reduce.write(item + ' ' + str(dict_result[item]))
    arquivo_reduce.write("\n")

  arquivo_reduce.close()
```

##6. Distância total percorrida por hora.

```python
def map_coluna(nome_arquivo_entrada, nome_arquivo_saida, indice_data: int, indice_distancia: int):
  """Lê a data/hora e a distância, extrai apenas a hora e gera o par (hora, distancia)."""
  arquivo_entrada = open(nome_arquivo_entrada, 'r')
  arquivo_saida = open(nome_arquivo_saida, 'w')
  primeira_linha = arquivo_entrada.readline()

  for linha in arquivo_entrada:
    if not linha.strip():
      continue
    colunas = linha.strip().split(',')
    data_hora = colunas[indice_data]
    distancia = colunas[indice_distancia]

    try:
      hora = data_hora.strip().split(' ')[1].split(':')[0]
      arquivo_saida.write(f"{hora}\t{distancia}\n")
    except (IndexError, AttributeError):
      continue

  arquivo_entrada.close()
  arquivo_saida.close()
```

```python
def shuffle(nome_arquivo_entrada, nome_arquivo_saida):
  """Agrupa as distâncias por hora convertendo o valor para float."""
  resultado_intermediario = defaultdict(list)

  with open(nome_arquivo_entrada) as arquivo_entrada:
    for linha in arquivo_entrada:
      if not linha.strip():
        continue
      chave, valor = linha.strip().split('\t')
      resultado_intermediario[chave].append(float(valor))

  resultado = dict(resultado_intermediario)
  arquivo_shuffle = open(nome_arquivo_saida, 'w')
  arquivo_shuffle.write(str(resultado))
  arquivo_shuffle.close()
```

```python
def reducer(nome_arquivo_entrada, nome_arquivo_saida, funcao_reduce):
  """Soma a distância total percorrida por hora."""
  arquivo = open(nome_arquivo_entrada, 'r')
  conteudo_entrada = arquivo.read()
  dict_entrada = eval(conteudo_entrada)
  arquivo.close()
  dict_result = {}

  for chave, lista in dict_entrada.items():
    dict_result[chave] = reduce(funcao_reduce, lista)

  arquivo_reduce = open(nome_arquivo_saida, 'w')

  for item in sorted(dict_result):
    arquivo_reduce.write(f"{item} {dict_result[item]:.2f}\n")

  arquivo_reduce.close()
```
